```python
# TASK 1: TEXT CLEANING & PREPROCESSING

import re
import nltk
from nltk.corpus import stopwords

nltk.download('stopwords')

raw_text = """Natural Language Processing (NLP) is growing extremely FAST!!!
It helps machines understand HUMAN language, and it's used in chatbots,
translators, and MANY other applications."""

print("RAW TEXT\n")
print(raw_text)

def clean_text(text):
    # 1. Lowercase
    text = text.lower()

    # 2. Remove punctuation + numbers (keep only alphabets and spaces)
    text = re.sub(r'[^a-z\s]', ' ', text)

    # 3. Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()

    # 4. Remove stopwords
    stop_words = set(stopwords.words('english'))
    tokens = text.split()
    filtered_tokens = [word for word in tokens if word not in stop_words]

    cleaned_text = " ".join(filtered_tokens)
    return cleaned_text

cleaned_text = clean_text(raw_text)

print("\nOutput\n")
print(cleaned_text)
```

```python
# TASK 2: TOKENIZATION

import nltk
from nltk.tokenize import word_tokenize, sent_tokenize

nltk.download('punkt')
nltk.download('punkt_tab')

print("\nPrevious Output")
print(cleaned_text)

# 1. WORD TOKENIZATION
word_tokens = word_tokenize(cleaned_text)

# 2. SENTENCE TOKENIZATION
sentence_tokens = sent_tokenize(cleaned_text)

print("\nWORD TOKENS\n")
print(word_tokens)

print("\nSENTENCE TOKENS\n")
print(sentence_tokens)
```

```python
# TASK 3: STEMMING & LEMMATIZATION

import nltk
from nltk.stem import PorterStemmer, WordNetLemmatizer

nltk.download('wordnet')
nltk.download('omw-1.4')

ps = PorterStemmer()
lemmatizer = WordNetLemmatizer()

# 1. STEMMING
stems = [ps.stem(word) for word in word_tokens]

# 2. LEMMATIZATION
lemmas = [lemmatizer.lemmatize(word) for word in word_tokens]

print("\nPrevious WORD TOKENS\n")
print(word_tokens)

print("\nSTEMMED WORDS")
print(stems)

print("\nLEMMATIZED WORDS")
print(lemmas)
```

```python
# TASK 4: N-GRAM GENERATION

from nltk import ngrams

print("Previous CLEANED TEXT:")
print(cleaned_text)

# 1. Tokenize cleaned text by simple split
tokens_for_ngrams = cleaned_text.split()

# 2. Generate bigrams (pairs of consecutive words)
bigrams = list(ngrams(tokens_for_ngrams, 2))

# 3. Generate trigrams (triplets of consecutive words)
trigrams = list(ngrams(tokens_for_ngrams, 3))

print("\nBIGRAMS (2-grams):")
print(bigrams)

print("\nTRIGRAMS (3-grams):")
print(trigrams)
```

```python
def semantic_search(query, num_docs=3):
    print("QUERY:", query)

    # split query into individual words
    tokens = query.lower().split()

    # keep only those words that are in the model vocabulary
    keywords = [w for w in tokens if w in model.vocab]

    if not keywords:
        print("No query words found in model vocabulary. Try different words.")
        return

    # now use Top2Vec's keyword search
    docs, scores, doc_ids = model.search_documents_by_keywords(
        keywords=keywords,
        num_docs=num_docs
    )

    for i, (doc, score, doc_id) in enumerate(zip(docs, scores, doc_ids)):
        print(f"\nResult {i+1}")
        print("Document ID:", doc_id)
        print("Score:", score)
        print("Snippet:")
        print(doc[:400], "...")
        print("-" * 80)
```
